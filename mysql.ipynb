```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
df = pd.read_parquet("./data/logs.parquet")
```

```python
engine = create_engine("mysql+mysqlconnector://admin:admin@localhost:3306/logs")
```

```python
filas = 10000
try:
    for inicio in range(0, len(df), filas):
        chunk = df.iloc[inicio:inicio+filas]
        chunk.to_sql(name="logs", con=engine, if_exists="append", index=False)
        print(f"Insertadas {inicio+filas} filas")
except Exception as e:
    print("Error al insertar:", e)
```

```python
pd.read_sql("SELECT count(*) FROM logs", con=engine)
```
